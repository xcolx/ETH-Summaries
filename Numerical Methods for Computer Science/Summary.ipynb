## Lösung quadratischer Gleichung

### mit Differentenqotient (Auslöschung)

Wir verwenden die Formel $p(x) = (x−c)(x−\frac{1}{c}) = x^2 −(c+\frac{1}{c})x+ 1$ in diesem Beispiel, da die Nullstellen so bekannt sind.

```python
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline

def mf(a,b,c):
    return ((-b + np.sqrt(b**2-4*a*c))/(2*a), (-b - np.sqrt(b**2-4*a*c))/(2*a))

cs = np.linspace(2, 1000, num=101, endpoint=True)

a = np.ones(cs.shape)
b = -(cs+1/cs)
c = np.ones(cs.shape)

y = mf(a,b,c)
y_plus = y[0]
y_minus = y[1]

y_plus_real = cs
y_minus_real = 1/cs

plus_err = abs(y_plus-y_plus_real)/y_plus_real
minus_err = abs(y_minus-y_minus_real)/y_minus_real

plt.plot(cs, plus_err, '+')
plt.plot(cs, minus_err, '.')

plt.semilogy()

plt.title("Auslöschung Beispiel")
plt.xlabel("c")
plt.ylabel("relativ error")
plt.legend(["Plus", "Minus"])
```

### Vieta Formel

Das Problem bisher ist, dass wir beim bestimmen einer der beiden Nullstellen Auslöschung haben. Diese können wir aber auch mit dem Satz von Vieta berechnen:

$x^2 + px + q = 0$

$x_1 + x_2 = -p$

$x_1 * x_2 = q$

$=> x_2 = b/x_1$


```python
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline

def mf(a,b,c):
    return ((-b + np.sqrt(b**2-4*a*c))/(2*a), (-b - np.sqrt(b**2-4*a*c))/(2*a))

cs = np.linspace(2, 1000, num=101, endpoint=True)

a = np.ones(cs.shape)
b = -(cs+1/cs)
c = np.ones(cs.shape)

y = mf(a,b,c)
y_plus = y[0]
y_minus = 1/y_plus # change

y_plus_real = cs
y_minus_real = 1/cs

plus_err = abs(y_plus-y_plus_real)/y_plus_real
minus_err = abs(y_minus-y_minus_real)/y_minus_real

plt.plot(cs, plus_err, '+')
plt.plot(cs, minus_err, '.')

plt.semilogy()

plt.title("Auslöschung Beispiel")
plt.xlabel("c")
plt.ylabel("relativ error")
plt.legend(["Plus", "Minus"])
```

Mit imaginärem Schritt: $f'(x_0) \approx Im(\frac{f(x_0 + ih)}{h})$ Fehler in $O(h^2)$ da Auslöschung vermieden

```python
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline

# Differenzialquotient
def diff_quot(f,x, h):
    return (f(x+h) - f(x))/h
    
# Imaginärer Schritt
def diff_im(f,x, h):
    return (np.imag(f(x+1.0j*h)))/h

x = 1.1 # where to evaluate
h0 = 0.5

datan = lambda x: 1/(1+x**2)
dsqrt = lambda x: 0.5*x**-0.5
dcos = lambda x: -np.sin(x)

funcs = [np.arctan, np.sqrt, np.exp, np.sin]
real_diffs = [datan, dsqrt, np.exp, np.cos]

h = h0/2**np.arange(60)
hs = h**2

for i in range(0,len(funcs)):
    plt.subplot(2,2,i+1)
    plt.title(funcs[i].__name__)
    plt.loglog()
    plt.xlim(10**(-17), 0)
    plt.ylim(10**(-17), 0)

    plt.plot(h, h, 'k--')
    plt.plot(h, hs, 'k-.')

    diff_quot_err = abs(diff_quot(funcs[i], x, h) - np.full(h.shape, real_diffs[i](x)))/np.full(h.shape, real_diffs[i](x))
    plt.plot(h, diff_quot_err, '.')

    diff_im_err = abs(diff_im(funcs[i], x, h) - np.full(h.shape, real_diffs[i](x)))/np.full(h.shape, real_diffs[i](x))
    plt.plot(h, diff_im_err, '.')

```

$$
    f'(x_0) \approx \frac{f(x_0 + h) - f(x_0 - h)}{2h}
$$

Fehler auch in $O(h^2)$

### Richardson

Wir wollen schneller Konvergieren so dass wir das Resultat (mit genügend kleinem Fehler) bekommen bevor wir Auslöschung haben.

Wir verwenden dazu wieder erst den Differenzenquotient zweiter Ordnung $\frac{f(x+h)-f(x-h)}{2h}$ wir nennen diesen $R_{l,0}$ für $h = \frac{h_0}{2^l}$

Wir können dann $R_{l,k}$ für weitere k berechnen, indem wir auf die vorherigen Resultate zurückgreiffen (DP) und verlieren dabei für jedes weitere k einen Fehlerterm:

$$
    R_{l,k} = \frac{4^kR_{l,k-1} - R_{l-1,k-1}}{4^k-1}
$$

Auf der Diagonale liegen dann die besten Werte im Richardson-Schema weil wir dort für kleinstmögliche Schrittweite maximal viele eliminierte Fehlerterme haben.

| ℓ \ k | 0        | 1        | 2        | 3        |
|------:|----------|----------|----------|----------|
| 0     | R₀,₀     |          |          |          |
| 1     | R₁,₀     | R₁,₁     |          |          |
| 2     | R₂,₀     | R₂,₁     | R₂,₂     |          |
| 3     | R₃,₀     | R₃,₁     | R₃,₂     | R₃,₃     |


## Polynominterpolation

Ziel ist es Funktionen $f$ zu gegebenen Datenpunkten $(x_i, y_i)$ ($x_i$ heissen Stützstellen oder Knoten) zu finden wobei $f(x_i) = y_i$

### Monombasis

Wenn wir k+1 Punkte haben können wir ein Polynom p(x) von Grad k eindeutig bestimmen, dass durch alle Punkte geht.

Die Koeffizienten lassen sich durch ein Gleichungssystem mithilfe der Vandermonde-Matrix berechnen:

$$ \begin{pmatrix} 1 & x_0 & \cdots & x_0^{n} \\ 1 & x_1 & \cdots & x_1^{n} \\ \vdots & \vdots & \ddots & \vdots \\ 1 & x_n & \cdots & x_n^{n} \end{pmatrix} \begin{pmatrix} \alpha_0 \\ \alpha_1 \\ \vdots \\ \alpha_n \end{pmatrix} = \begin{pmatrix} y_0 \\ y_1 \\ \vdots \\ y_n \end{pmatrix} $$

Jedoch ist die Vandermonde-Matrix sehr schlecht konditioniert. Auserdem gibt es Probleme bei der Auswertung welche sich jedoch durch das Horner-Schema beheben lassen:

$$
p(x) = \bigl(x \cdots x\bigl(x(\alpha_n x + \alpha_{n-1}) + \cdots + \alpha_1\bigr) + \alpha_0\bigr)
$$

### Newton-Basis

Diese Basis ermöglicht es im gegensatz zu Monomen mehr Punkte hinzuzufügen ohne alles nochmal neu berechnen zu müssen.

Newton-Basis: $N_0 = 1, N_1 = x-x_0, N_2 = (x-x_0)(x-x_1), ...$

Die Koeffizienten des Newton Polynoms: $p_n(x) = \beta_0N_0(x) + \beta_1N_1(x) + ...$ können wir entweder wieder durch ein lineares Gleichungssystem bestimmen oder durch die Methode der Dividierten Differenzen:

$$
y[x_i]=y_i,\quad y[x_i,\ldots,x_{i+k}]=\frac{y[x_{i+1},\ldots,x_{i+k}]-y[x_i,\ldots,x_{i+k-1}]}{x_{i+k}-x_i}
$$


$x_0:\ y[x_0]\to y[x_0,x_1]\to y[x_0,x_1,x_2]\to y[x_0,x_1,x_2,x_3]$

$x_1:\ y[x_1]\to y[x_1,x_2]\to y[x_1,x_2,x_3]$

$x_2:\ y[x_2]\to y[x_2,x_3]$

$x_3:\ y[x_3]$

```python
# Beispielcode Berechnung der Newton Koeffizienten mithilfe der Dividirten Differenzen
def divdiff(x, y):
    n = y.size
    T = np.zeros((n, n))
    T[:,0] = y
    for level in range(1, n):
        for i in range(n-level):
            T[i, level] = (T[i+1,level-1] - T[i,level-1]) / (x[i+level]-x[i])
    return T[0,:]
```

### Lagrange- und baryzentrische Interpolationsformeln

Wir definieren die **Lagrange-Polynome** wie folgt:

$l_i(x) = \prod_{j=0, j \neq i}^n \frac{x-x_j}{x_i-x_j}$

Damit haben wir die Koeffizienten direkt schon gegeben indem wir die y Werte nehmen:

$p(x) = \sum_{j=0}^n y_il_i(x)$

Das funktioniert in der Theorie schon. Jedoch haben wir in der Praxis die Probleme, dass wir 1. $O(n^2)$ Zeit brauchen um einen Punkt auszuwerten und 2. das ganze Nummerisch instabil ist, da bei einem grossen n der Zähler und Nenner der Lagrange Polynome sehr gross/klein wird.

Deshalb benutzen wir in realität die **baryzentrische Form**:

Wir definieren $\lambda_k = \prod_{j \neq k} \frac{1}{x_k - x_j}$

Wir können dan durch längeres Umformen (Skript) zeigen dass $p(x) = \frac{\sum_{k=0}^n \frac{\lambda_k}{x-x_k} y_k}{\sum_{k=0}^n \frac{\lambda_k}{x-x_k}}$

Die neuen Vorteile:
- Laufzeit: $O(n^2)$ einmalig für Gewichte berechnen, Auswertung und neue Punkte hinzufügen kostet $O(n)$
- Stabilität: Nummerisch stabil
- Flexibilität: Die Gewichte hängen nur von den Stützpunkten (x-en) ab. Wenn diese gleich bleiben und sich die Messwerte ändern können wir immer noch in $O(n)$ auswerten

Für den **Fehler** definieren wir die Lebesgue-Konstante $\Lambda_n$ mit welchem wir abschätzen. Das grosse **Problem** was auftreten kann ist das sogenannte **Runge-Phänomen**. (tritt zum Beispiel bei $f(x) = \frac{1}{1+x^2}$ auf) Dabei treten starke Oszillationen an den Randbereichen auf.

Deshalb: Verwende für eine Interpolation hohen Grades niemals äquidistant verteilte Stützstellen!

### Chebyshev-Interpolation

Wir wollen bessere Stützstellen finden so dass wir das Runge-Phänomen vermeiden.

Dazu definieren wir erst einmal die **Chebyshev-Polynome**:

erster Art: $T_n(x) = cos(n*arccos(x)), x \in [-1,1]$

zweiter Art: $U_n(x) = \frac{sin((n+1)arccos(x))}{sin(arccos(x))}, x \in [-1,1]$

mit der Eigenschaft: $T_0(x) = 1, T_1(x) = x, T_{n+1}(x) = 2xT_n(x) - T_{n-1}(x)$

Die n+1 **Chebyshev-Knoten** (Nullstellen von $T_{n+1}(x)$) definieren wir dann wie folgt:

$x_k = a + \frac{1}{2}(b-a)(cos(\frac{2k+1}{2(n+1)}\pi)+1), k = 0, ..., n$

Die **Chebyshev-Abszissen** (Extrema von $T_n(x)$) so:

$x_k = a+\frac{1}{2}(b-a)(cos(\frac{k}{n}\pi)+1), k=0,...,n$

(Wir nehmen nur k = 1,...,n-1 wenn wir die Endpunkte a und b ausschliessen wollen.)

.

Es stellt sich heraus, dass die Chebyshev-Knoten die best mögliche Wahl an Punkten für die Lagrange Methode ist. Aber wir verwenden in der Praxis die Chebyshev-Abszissen (close enough und nummerisch stabiler).

.

Wir können die Chebyshev-Polynome auch direkt als Basis verwenden: $p(x) = c_0T_0(x)+c_1T_1(x)+...+c_nT_n(x)$

Die Chebyshev-Polynome sind orthogonal bezüglich des Skalarprodukts (und des diskreten Skalarprodukts im Raum der Polynome Grad <= n).

Aus diesem Grund, können wir relativ simpel die Koeffizienten bestimmen: $c_k = \frac{\sum_{i=0}^{n} y_i\, T_k(x_i)}{\sum_{i=0}^{n} T_k(x_i)^2}$. Wir können aber auch FFT benutzen (kommt später).

.

Rechenaufwand zum Bestimmen des Interpolationspolynoms:

Direkt: $O((n+1)^2)$

mittels FFT: $O(n*log(n))$



#### Clenshaw-Algorithmus

Dieser Algorithms kann verwendet werden um in $O(n)$ nummerisch stabil, das Chebyshev-Polynom auszuwerten. 

Wir setzen $d_{n+2} = d_{n+1} = 0$

$d_k = c_k + (2x)d_{k+1} - d_{k+2}$

Wir erhalten: $p(x) = d_0 - xd_1 = \frac{1}{2} (d_0 - d_2)$.

## Trigonometrische Interpolation

Wir wollen die trigonometrische Interpolation verwenden, da sie das Runge Phänomen vermeidet.

### Fourier Reihe

Die Fourier Reihe ist definiert als:

$f(t) \sim \frac{a_0}{2} + \sum_{k=1}^{\infin} (a_k \cos(2 \pi k t) + b_k \sin(2 \pi k t))$

oder auch

$f(t) \sim \sum_{k=-\infty}^{\infty} c_k\, e^{2\pi i k t}$

in der komplexen Form.

Das sind Reihen (also unendliche Summen) wie in der Analysis definiert.

### Trigonometrisches Polynom

Wir definieren das trigonometrische Polynom als endliche Teilsumme der Fourier Riehe:

$p_m(t) = \frac{a_0}{2} + \sum_{k=1}^m (a_k \cos(2 \pi k t) + b_k \sin(2 \pi k t))$

oder 

$p_m(t) = \sum_{k=-m}^m c_k e^{2 \pi ikt}$

in der komplexen Form.

Mit N Funktionswerten (samples) können wir maximal $\deg(p) = m = \left \lfloor \frac{N-1}{2} \right \rfloor$ viele Frequenzen bestimmen.

Wir nehmen im generellen immer äquidistante Punkte und lassen die Randpunkte weg. 

#### Orthogonalität

"Intuition: Two different frequencies are like perpendicular directions in an abstract space. Over a full cycle their ups and downs cancel, so they carry no information about each other. This independence means we can measure each frequency’s contribution separately, without contamination."

### DFT

Mithilfe der Diskreten Fourier Transformation können wir aus N (endlich vielen) Messpunkten das Polynom $p(t) = \sum_{-m}^m c_ke^{2\pi ikt}$ bestimmen, wobei $m = \left \lfloor \frac{N-1}{2} \right \rfloor$ die maximale Anzahl Frequenzen ist, die wir mit N Datenpunkten bestimmen können.

Dazu verwenden wir die **Einheitswurzeln** $\omega_N = e^{-2\pi i /N}$. Mit ihnen definieren wir die Fourier matrix $F_N[i,j] = \omega_N^{ij}$

Beispiel $F_4$:

$F_4 = \begin{bmatrix} 1&1&1&1 \\ 1&\omega_4&\omega_4^2&\omega_4^3 \\ 1&\omega_4^2&\omega_4^4&\omega_4^6 \\ 1&\omega_4^3&\omega_4^6&\omega_4^9 \end{bmatrix}$

Mithilfe dieser Matrix und den Messpunkten y können wir die Koeffizienten c bestimmen indem wir $c = F_N * y$ rechnen.

Mithilfe der **Inversen DFT** bekommen wir aus den Koeffizienten c dann wieder die Messpunkte:

$y = \frac{1}{N} F_N^H c$

Die Laufzeit ist $O(N^2)$, mit der FFT geht es aber in $O(N\log(N))$.

#### Aliasing

Aliasing ist der Effekt, dass bei zu grober Abtastung hohe Frequenzen als falsche niedrige Frequenzen erscheinen.
Beim Abtasten eines kontinuierlichen Signals gehen Informationen über hohe Frequenzen verloren.
Diese hohen Frequenzen werden in der DFT „umgefaltet“ und erscheinen als niedrigere Frequenzen.
Ursache ist eine Verletzung des Nyquist-Kriteriums (Abtastrate < 2 × maximale Frequenz).
Aliasing führt zu nicht reparierbaren Fehlern in Rekonstruktion und Interpolation.

### FFT

Die FFT ist von Input/Resultat genau das gleiche wie die DFT aber eine schnellere Implementerung davon.

Sie nutzt die spezielle Struktur der DFT-Matrix (Divide-and-Conquer, Symmetrien), um redundante Berechnungen zu vermeiden und ist daher um einiges Effizienter ($O(N\log(N))$) als DFT.

Alle Libraries benutzen deshalb immer FFT.

### Python Implementierung

In Python können wir die Funktionen **np.fft.fft** und **np.fft.ifft** benutzen. 

Um die Koeffizienten c zu erhalten rechnen wir c = (1/n) * np.fft.fft(y)

Um dann wieder die Messpunkte y aus den Koeffizienten c zu erhalten rechnen wir y = N * np.fft.ifft(c)

Wichtiger Hinweis (war in den Lösungen gewisser CE Aufgaben): Wenn wir nur reelle Messwerte haben ist c = (1/n) * np.fft.fft(y) = np.fft.ifft

.

**np.fft.fftshift**: Wenn wir np.fft.fft benutzen, bekommen wir die Frequenzwerte in einem Array [positive low - positive high - negative high - negative low].
fftshift verschiebt die Nullfrequenz (DC-Anteil) in die Mitte des Spektrums. Dadurch werden negative Frequenzen links und positive rechts symmetrisch um die Mitte angeordnet, was die Visualisierung erleichtert.

### Powerspektrum

Das Powerspektrum beschreibt die Verteilung der Signalenergie über die Frequenzen.
Es wird aus den Fourier-Koeffizienten $c_k$ berechnet als $\frac{1}{N} * |c_k|^2$
Jeder Wert $|c_k|^2$ gibt an, wie stark die zugehörige Frequenz zum Signal beiträgt.
Bei reellen Signalen ist das Powerspektrum symmetrisch, da positive und negative Frequenzen komplex konjugiert sind.
Für die Darstellung wird häufig fftshift verwendet, um das Spektrum um die Nullfrequenz zu zentrieren.

### Frequency Filtering

Idee:

$y \xrightarrow{\text{DFT}} c \xrightarrow{\text{Filtering}} c' \xrightarrow{\text{Inverse DFT}} y'$

Wir filtern die Frequenzen direkt in der Frequencydomain indem wir zum Beispiel:
- zu hohe Frequenzen entfernen (Low-pass filter): denoising, image blurring, data compression
- zu tiefe Frequenzen entfernen (High-pass filter): emphasize edges and abrupt changes
- einen spezifischen Frequenzbereich behalten (Band filters)

```python
# Beispielcode für einen low-pass filter

import numpy as np
# Forward transform (time→ frequency)
c = np.fft.fft(y)
# Design a simple low-pass filter
cutoff = N // 8
mask = np.zeros(N)
mask[:cutoff] = 1
mask[-cutoff:] = 1
# Apply filter in frequency domain
c_filtered = c * mask
# Inverse transform (frequency→ time)
y_filtered = np.fft.ifft(c_filtered)
```

### Auswertung

#### Zero Padding

Für gerades n können wir die Interpolationsauswertung wie folgt machen:
Wir erweitern in der Mitte der Koeffizienten c mit 0en. Das entspricht physikalisch dem hinzufügen von hohen Frequenzen mit Anteil 0.
Mit ifft erhalten wir dann die Funktionswerte y auf einem feineren Gitter interpoliert.

```python
m = n // 2

c = np.fft.fft(y) / n 

a = np.zeros(N, dtype=complex)
a[:m] = c[:m]
a[N-m:] = c[m:]

v = np.real(np.fft.ifft(a) * N)
```

#### Ansonsten: Ausrechnen

Wenn man kein feineres äquidistantes Gitter braucht wie beim Zero Padding, oder n nicht gerade ist, bleibt nichts anderes übrig, als das trigonometrischen Interpolationspolynom für jeden Punkt auszuwerten:

$y(t)= \sum_{k=-m}^{m-1} c_k\, e^{2\pi i k t}$

### Fehler

Falls die Funktion k 1-periodisch ist und die Summe der Fourier-Koeffizienten $c_k$ absolut konvergiert, können wir abschätzen
$$
| p_N(x) - f(x) | \leq 2 \sum_{|k| \geq N/2} |c_k|
$$

Falls f 1-periodisch ist, die maximale Frequenz m ist und N > 2m, können wir genau approximieren: Wenn $c_k = 0$ für $|k| > m$ dann gilt $p_N(x) = f(x)$

#### Gibbs Phänomen

Das Gibbs-Phänomen ist ein dauerhaftes Überschwingen der Fourier-Approximation an Sprungstellen, das auch für $N \to \infty$ nicht verschwindet.
Bei Funktionen mit Sprüngen erzeugt die Fourier-Reihe in der Nähe der Unstetigkeit Oszillationen.
Die maximale Überschwingung bleibt bei ca. 9 % der Sprunghöhe.
Mit wachsendem N wird die Überschwingungszone schmaler, aber nicht niedriger.

## Stückweise Interpolation und Splines

Die Idee: Wir zerlegen das ganze Intervall in kleine Intervalle und schätzen lokal das Polynom mit einem Polynom von geringem Grad ab.

Dazu teilen wir das Intervall mit Knoten $a = x_0 < x_1 < ... < x_N = b$ an denen wir die Funtionswerte kennen.

h beschreibt die Subintervalllänge: $h = max(x_j-x_{j-1})$

### Lineare stückweise Interpolation

$s_j(x) = y_{j-1} \frac{x_j - x}{h_j} + y_j \frac{x - x_{j-1}}{h_j}$

Anschaulich ist das die simple Idee die Knoten einfach mit einer Gerade zu verbinden.

### Kubische Hermite-Interpolation

Die Idee ist mit einem kubischen Polynom (Grad 3) abzuschätzen, da das genauer ist.

Dafür brauchen wir aber zusätzlich die Ableitungen an den Knoten. Wir setzten voraus, dass Funktionswert und Ableitung an den Knoten exakt stimmen.

Dadurch ergibt sich ein Gleichungssystem mit vier Gleichungen: $s(x_{j-1}) = y_{j-1}, \; s(x_j) = y_j, \; s'(x_{j-1}) = c_{j-1}, \; s'(x_j) = c_j$

Da ein Polynom 3. Grades durch 4 Koeffizienten bestimmt wird, können wir das ganze mit einm Gleichungssystem bestimmen. Effizienter und nummerisch stabiler geht die Lösung aber so:

$p_j(x) = y_{j-1} \varphi(\frac{x - x_{j-1}}{h_j}) + y_j \varphi(\frac{x_j - x}{h_j})$

$q_j(x) = c_{j-1} h_j \psi(\frac{x - x_{j-1}}{h_j}) - c_j h_j \psi(\frac{x_j - x}{h_j})$

$s_j(x) = p_j(x) + q_j(x)$

mit $\varphi(t) = t^2(3 - 2t)$ und $\psi(t) = t^2(t-1)$

Wenn nur die Funktionswerte aber nicht die Ableitungen gegeben sind, haben wir die Freiheit diese selbst zu bestimmen mit verschiedenen Methoden z.B. Akima oder PCHIP.

### Splines

Motivation: Runge Problem wird "behoben"

Bei Splines von Grad d (Ordnung d+1) schätzen wir die Subintervalle mit Polynomen von Grad höchstens d ab. (Lineare stückweise Interpolation ist ein 1-d Spline.)

Wir brauchen also für Splines von grad d (d-1) Ableitungen in den Knoten.

Der Freiheitsgrad der Splines ist $\dim \mathcal{S}_{d,G} = (Koeffizienten)-(Bedingungen) = (d+1)N - (N-1)d = N + d$

.

Da wir für Grad 3 also kubische Splines N+1 Funktionswerte (Gleichungen) gegeben haben und N+3 Freiheitsgrade, brauchen wir nochmals 2 Bedignungen (Gleichungen). Es gibt verschiedene Arten diese zu definieren:

- **vollständig (=eingespannt):** $s'(a)=c_a$ und $s'(b)=c_b$ ("Ableigungen an den Randpunkten gegeben")
- **natürlich:** $s''(a) = s''(b) = 0$
- **periodisch:** $s'(a) = s'(b)$ und $s''(a) = s''(b)$ nur sinnvoll wenn $y_a = y_b$ 
- **not-a-knot (=NaK):** $s'''$ ist stetig in $x_1$ und $x_{N-1}$ ("Erstes und zweitens Intervall bilden ein Polynom")

## Quadratur

Bei der Quadratur geht es darum, Integrale nummerisch zu lösen.

Dazu haben wir im generellen $Q_n(f) = \sum_{i=1}^n w_i f(c_i)$

für $w_i$ Gewichte und $c_i$ Stützstellen/Knoten

**Konvergenz**
- algebraisch: $E(n) = O(\frac{1}{n^p})$
- exponentiell: $E(n) = O(q^n)$

**Ordnung**
Eine Quadraturregel hat Ordung n+1 wenn sie Polynome von Grad n exakt löst und bei n+1 nur approximiert.

**Symmetrie**
Eine Quadraturregel auf [-1,1] ist symmetrisch wenn $w_i = w_{n+1-i}$ und $c_i = -c_{n+1-i}$. Die Ordnung von symmetrischen Regeln ist immer gerade.

**Methoden**
- h-Methode: Schrittweite h muss reduziert werden um bessere Resultate zu erzielen
- p-Methode: Polynomgrad p muss erhöht werden um bessere Resultate zu erzielen

### Newton-Cotes Regeln

Newton-Cotes Quadraturregeln approximieren das Integral durch **äquidistante Stützstellen** und **Polynominterpolation**

#### Einfache Regeln

**Mittelpunkt-Regel**

$Q(f,a,b) = (b-a) f(\frac{a+b}{2})$

Ordnung: 2

**Trapez-Regel**

$Q(f,a,b) = \frac{b-a}{2} (f(a) + f(b))$

Ordnung: 2

**Simpson-Regel**

$Q(f,a,b) = \frac{b-a}{6} (f(a) + 4f(\frac{a+b}{2}) + f(b))$

Ordnung: 4

##### Code Beispiele:

Setup

```python
import numpy as np
from scipy.integrate import quad, trapezoid, simpson
import random
# Function to integrate
def f(x):
    return np.exp(-x**2)
# Integration bounds
a, b = 0, 1
```

Manual Implementations

```python
# =======================================================
# (a) Midpoint Rule
# =======================================================
def midpoint_rule(f, a, b, n):
    """Composite Midpoint Rule with n subintervals"""
    h = (b - a) / n
    mids = a + (np.arange(n) + 0.5) * h
    return h * np.sum(f(mids))


# =======================================================
# (b) Trapezoidal Rule
# =======================================================
def trapezoidal_rule(f, a, b, n):
    """Composite Trapezoidal Rule"""
    h = (b - a) / n
    x = np.linspace(a, b, n + 1)
    return (h / 2) * (f(x[0]) + 2 * np.sum(f(x[1:-1])) + f(x[-1]))


# =======================================================
# (c) Simpson’s Rule 
# =======================================================
def simpson_rule(f, a, b, n):
    """Composite Simpson's Rule (requires n even)"""
    if n % 2 != 0:
        n += 1
    h = (b - a) / n
    x = np.linspace(a, b, n + 1)
    return (h / 3) * (f(x[0]) + 4 * np.sum(f(x[1:-1:2])) + 2 * np.sum(f(x[2:-2:2])) + f(x[-1]))


# =======================================================
# (d) Monte Carlo
# =======================================================
def monte_carlo(f, a, b, n):
    """Monte Carlo"""
    s = 0.0
    for _ in range(n):
        x = random.uniform(a, b)
        s += f(x)
    return (b - a) * s / n
```

Evaluate and compare

```python
# Number of subintervals
n = 10
I_mid = midpoint_rule(f, a, b, n)
I_trap = trapezoidal_rule(f, a, b, n)
I_simp = simpson_rule(f, a, b, n)
I_monte = monte_carlo(f, a, b, n)
print("Midpoint rule: ", I_mid)
print("Trapezoidal rule:", I_trap)
print("Simpson rule: ", I_simp)
print("Monte Carlo: ", I_monte)
```

Using scipy.integrate

```python
# =======================================================
# 1. Adaptive high-precision integration (reference value)
# =======================================================
I_true, err = quad(f, a, b)
print("True value (quad):", I_true)


# =======================================================
# 2. Uniform-grid integration (fixed subintervals)
# =======================================================
x = np.linspace(a, b, 101)
I_trap_scipy = trapezoid(f(x), x)
I_simp_scipy = simpson(f(x), x)
print("Trapezoidal (SciPy):", I_trap_scipy)
print("Simpson (SciPy): ", I_simp_scipy)
```

#### Summierte Regeln

Wir können die drei simplen Regeln auch auf N kleinere Intervalle anwenden und können somit global besser approximieren. Die Ordnung bleibt aber gleich.

**Summiere Mittelpunkt-Regel**

$Q^n (f) = h \sum_{k=0}^{n-1} f(\frac{x_k + x_{k+1}}{2})$

**Summierte Trapez-Regel**

$Q^n(f) = \frac{h}{2} (f(a) + 2 \sum_{i=1}^{n-1} f(x_i) + f(b))$

**Summierte Simpson-Regel**

$Q^n(f) = \frac{h}{6} \sum_{i=0}^{n-1} (f(x_i) + 4 f(\frac{x_i + x_{i+1}}{2}) + f(x_{i+1}))$

#### Romberg (Richardson Extrapolation)

Romberg basiert auf der Extrapolation der Trapezregel und ist somit eine Strategie zur verbesserung der Ordnung und Approximation aber keine egienständige Regel.

Bei algebraischer Konvergenz haben wir
$$
    E(h/2) \approx \frac{E(h)}{2^p}
$$

Wir können das verwenden, um den genauen Wert zu **extrapolieren**, ohne die Funktion noch einmal auswerten zu müssen. Wenn wir wissen, dass $Q_h \approx I + E(h)$ und $Q_{h/2} \approx I + E(h)/4$, dann können wir eine Abschätzung für den Fehler erhalten und diesen von $Q_{h/2}$ abziehen.

Die **Romberg Integration** verwendet dieselbe Idee wie die Konvergenzbeschleunigung nach Richardson aus Kapitel 1. Da die Trapezregel symmetrisch ist und nur gerade Potenzen von $h$ im Fehler hat, können wir das Schema hier anwenden.

Wir haben $R_{l,0} = T(2^l)$ wobei $T(n)$ die approximation mit $n$ Punkten ist. Wir starten bei $l=1$ und berechnen die Werte durch DP mit
$$
    R_{l,k} = \frac{4^kR_{l,k-1} - R_{l-1,k-1}}{4^k-1}
$$

Wir können dabei für $R_{l+1,0}$ die alten Punkte von $R_{l,0}$ wiederverwenden. Romberg funktioniert sehr gut, wenn die Funktion sehr glatt ist und keine Rundungsfehler $h$ dominieren.

### Gauss

Gauss-Quadratur ist eine optimale Quadraturregel, die Polynome bis Grad $2n−1$ mit $n$ Stützstellen exakt integriert. Sie haben also **Ordnung** $2n$

Statt äquidistanter Punkte (wie bei Newton–Cotes) wählt man optimal platzierte Knoten und Gewichte.
Die Knoten sind Nullstellen orthogonaler Polynome, wodurch die Genauigkeit maximal wird.

Es gibt zwei bekannte warten die wir besprochen haben:

**Legendre-Polynome** (Standard Fall, wird auch nur Gauss genannt):
- $w(x) = 1$ (Gewichtsfunktion nicht Quadraturgewichte)
- auf [-1, 1]
- Knoten sind Nullstellen von den Legendre-Polynomen
- Für Gewichte gibt es eine Formel die im Skript erwähnt wird
- Man kann für die Berechnung auch den Algorithmus von Golub und Welsch benutzen (Code im Skript). Aber sehr ungenau!
- In der Praxis berechnen (für n Punkte): (Knoten, Gewichte) = np.polynomial.legendre.leggauss(n) = scipy.special.roots_legendre(n)

**Hermite-Polynome**
- $w(x) = e^{-x^2}$
- auf $(-\infin, \infin)$
- Knoten sind Nullstellen von den Hermite-Polynomen
- In der Praxis berechnen (für n Punkte): (Knoten, Gewichte) = np.polynomial.hermite.hermgauss(n) = scipy.special.roots_hermite(n)

Es gibt auch noch andere Varianten wie die **Radau-Quadratur** und die **Lobatto-Quadratur** die im Skript nur kurz erwähnt sind. Diese haben aber eine geringere Ordnung. Da einer oder beide Endpunkte Knoten sind. Lösen kann man das mit scipy.special.roots_jacobi

### Clenshaw-Curtis

Clenshaw-Curtis verwendet die Chebyshev-Abszissa als Knoten. Es verwendet FFT und läuft für $2s$ Knoten in $O(s\log(s))$, konvergiert auch exponentiell, aber etwas langsamer als Gauss. Der Code dazu findet sich alles im Skript.

### Monte-Calro

Wir approximieren 

$\int_a^b f(x)dx \approx \frac{b-a}{N} \sum_{i=1}^N f(z_i)$

wobei $z_i$ zufällig gleichverteilte Punkte aus [a, b] sind.

**Vorteile/Nachteile:** Die Monte-Carlo Quadratur ist nützlich, da sie auch für hohe Dimensionen oder unglatte Funktionen funktioniert und eine kurze Laufzeit hat. Sie konvergiert nur langsam mit $O(N^{-1/2})$, aber dafür unabhängig von der Dimension. Es gibt kein Runge-Phänomen. Sie ist probabilistisch. 

#### Varianz Reduktion

Es gibt zwei Arten wie die Varianz reduziert und somit die Genauigkeit dieser Methode verbessert werden kann.

**Control Variates**

Wir nehmen eine Funktion $\phi(t)$ von der das Integral bekannt ist und die möglichst ähnlich zu $f(x)$ ist.

Dann integrieren wir

$\int_0^1 f(t)dt = \int_0^1 f(t) - \phi(t)dt + \int_0^1 \phi(t)dt$

Eine Wahl die oft geht dafür ist ein Taylor-Polynom von geringem Grad.

**Importance Sampling**

Dafür wählen wir die Punkte bei welchen wir $f(x)$ auswerten nicht zufällig gleichverteilt, sondern mit einer Dichtefunktion.

$\frac{1}{N} \sum_{i=1}^N \frac{f(x_i)}{g(x_i)}$

Wir wollen $g$ so wählen, dass die Werte, wo $f(x)$ gross ist, eine höhere Wahrscheinlichkeit bekommen.

### Adaptiv

Statt überall gleich fein zu rechnen, wird nur dort verfeinert, wo die Funktion schwierig ist (z.B. starke Krümmung, Singularitäten). Die Entscheidung zur Verfeinerung basiert auf einer Fehlerabschätzung. Adaptivität erhöht die Effizienz, ohne die zugrunde liegende Methode zu ändern.

## Nullstellensuche / Fixpunktsuche

Bei Linearen Systemen können wir Nullpunkte via Geichungssystem (Elimination) finden. Bei nicht linearen systemen ist das komplizierter. Dort müssen wir sehr schnell schon nummerische Methoden anwenden.

**Konvergenz**

Die Konvergenzordnung p > 0 ist folgendermassen definiert:

$\lim_{k \to \infty} \frac{|e^{(k+1)}|}{|e^{(k)}|^p} = C$, für ein $C>0$

Konvergenzarten:
- p = 1: linear
- p > 1: superlinear
- p = 2: quadratisch


**Fixpunkt vs Nullpunkt**

Wir können ein Problem eine Nullstelle zu finden immer in ein Porblem umbauen einen Fixpunkt zu finden und umgekehrt.

### Fixpunktiteration

Für ein Fixpunkt problem:
$x_{k+1} = f(x_k)$

Geometrisch entspricht das Auswerten -> auf x-y Gerade Projezieren -> Auswerten am neuen Punkt (repeat)

Konvergenzbedingung: $abs(f'(x^*)) < 1$

Wenn diese Bedingung nicht gegeben ist haben wir immer Divergenz. Ansonsten ist es Startpunkt abhängig.

Konvergenzordnung ist 1 und im fall $f'(x) = 0$ sogar 2.

```python
def fixpointiteration(f, x0, tol, Nmax):

    current = x0

    for _ in range(Nmax):
        last = current
        current = f(last)

        if abs(current-last) < tol:
            break

    return current
```

### Bisektionsverfahren

Wird benutzt für ein Nullstellenproblem.

Man benutzt den Zwischenwertsatz aus der Analysis: solange $f(a)*f(b) < 0$ ist eine Nullstelle zwischen a und b.

Man bestimmt den Mittelpunkt $m = \frac{a+b}{2}$ und fährt dann je nach Vorzeichen mit dem Intervall [a, m] oder [m, b] fort.

```python
def bisektion(f, a, b, Nmax):
    assert(f(a)*f(b) <= 0)

    k = 0
    while k < Nmax:
        m = (a+b)/2
        if (f(a)*f(m) < 0):
            b = m
        else:
            a = m
        k += 1

    return m
```

### Newton-Verfahren

Für Nullstellenproblem.

$x_{k+1} = x_k - \frac{F(x_k)}{F'(x_k)}$

Geometrische Interpretation: Tangente an Punkt legen und bei x = Schnittstelle Tangente-x-Achse weitermachen.

```python
def newton(f, df, x0, atol, rtol, Nmax):
    x = x0

    k = 0
    while k < Nmax:
        
        x_new = x - f(x)/df(x)
        k += 1

        if abs(x-x_new) <= atol + rtol*abs(x):
            break
        x = x_new

    return x_new
```

### Sekantenverfahren

Für Nullstellensuche.

Basiert auf den Newton-Verfahren. Benötigt aber die Ableitung nicht, da wir eine Sekante nehmen durch zwei Punkte.

$x_{k+1} = x_k - \frac{f(x_k)}{q_k}$

Wobei $q_k = \frac{f(x_k) - f(x_{k-1})}{x_k - x_{k-1}}$ die Sekantensteigung.

```python
def sekanten(f, x0, x1, atol, rtol, Nmax):

    for _ in range(Nmax):
        s = f(x1)*(x1-x0)/(f(x1)-f(x0))
        x0 = x1; x1 -= s

        if abs(s)<atol + rtol*abs(x1):
            break

    return x1
```

### n-Dimensionales Newton-Verfahren

Die Idee ist genau die selbe wie im 1-dimensionalen aber man benutzt jetzt als ableitung die Jaboi-Matrix DF(x).

$x_{k+1} = x_k - DF(x_k)^{-1} F(x_k)$

#### 2-D Beispiel

```python
import numpy as np
from scipy.linalg import solve, norm

F = lambda x: np.array([x[0]**2 - x[1]**4, x[0] - x[1]**3])

DF = lambda x: np.array([
                            [2*x[0], -4*x[1]**3],
                            [1.0, -3*x[1]**2]
                        ])

def newton_2d(F, DF, x0, tol=1e-10, Nmax=50):
    x = x0.copy()

    for _ in range(Nmax):
        s = solve(DF(x), F(x))
        x = x - s
        if norm(s) < tol:
            break

    return x
```

### Gedämpftes Newton-Verfahren

Manchmal "springt Newton zu weit" und wir verfehlen die Nullstelle. Deshalb können wir versuchen die Schritte zu dämpfen indem wir einen Dämpfungsfaktor $\alpha \in (0, 1]$ einführen.

$x_{k+1} = x_k - \alpha_k * \frac{f(x_k)}{f'(x_k)}$

Dieser Parameter kann fix sein oder sich den Gegebenheiten anpassen in jedem Schritt.

### Quasi-Newton Verfahren (Broyden)

Was wenn DF(x) zu teuer oder nicht zur Verfügung steht? Wir wollen also DF (= J) laufend approximieren.

Broyden ist eine Art wie man das macht. (Die einzige im skript beschriebene Umsetzung von Quasi-Newton)

$J_{k+1} := J_k + \frac{F\!\left(x^{(k+1)}\right)\,\left(\Delta x^{(k)}\right)^{T}}{\left\lVert \Delta x^{(k)} \right\rVert_2^{2}}$


Die Implementierung des Broydens-Verfahrens beruht auf dem Sherman-Morrison-Woodbury Update (SMW):

$J_{k+1}^{-1}=\left(I-\frac{J_k^{-1}F\!\left(x^{(k+1)}\right)(\Delta x^{(k)})^{T}}{\lVert\Delta x^{(k)}\rVert_2^{2}+(\Delta x^{(k)})^{T}J_k^{-1}F\!\left(x^{(k+1)}\right)}\right)J_k^{-1}$

#### Beispielcode

```python
import numpy as np
from scipy.linalg import lu_factor, lu_solve

def broyden_smw(x0, F, J0, tol=1e-12, maxit=20):
    x = x0.astype(float).copy()
    n = x.size

    # Start: J0^{-1} einmal berechnen (über LU; numerisch stabiler als inv)
    lup = lu_factor(J0)
    Jinv = lu_solve(lup, np.eye(n))

    fx = F(x)

    for k in range(maxit):
        # Quasi-Newton Schritt: Δx_k = - J_k^{-1} F(x_k)
        s = Jinv @ fx
        x_new = x - s

        # Abbruch (wie im Skript: über die Korrekturgröße)
        if np.linalg.norm(s, 2) <= tol:
            return x_new, k + 1

        f_new = F(x_new)

        # SMW-Update für J_{k+1}^{-1} gemäß Skriptformel
        u = Jinv @ f_new
        denom = (s @ s) + (s @ u)     # ||Δx||^2 + (Δx)^T J^{-1} F(x_{k+1})

        # (I - u s^T / denom) Jinv
        Jinv = (np.eye(n) - np.outer(u, s) / denom) @ Jinv

        x, fx = x_new, f_new

    return x, maxit
```

### QR-Zerlegung

Die QR-Zerlegung zerlegt eine Matrix $A$ in das Produkt $A=QR$, wobei die Spalten von $Q$ orthonormal sind und $R$ eine obere Dreiecksmatrix ist. Wir berechnen sie durch Gram-Schmidt, Householder-Transformationen oder Givens-Rotationen. In der Numerik wird sie vor allem zur stabilen Lösung von Least-Squares-Problemen verwendet.

#### Gram-Schmidt

$q_1 = \frac{a_1}{\lVert a_1 \rVert}$

for $k = 2,\dots,n$:

- $q_k' = a_k - \sum_{i=1}^{k-1} (a_k^{T} q_i)\, q_i$

- $q_k = \frac{q_k'}{\lVert q_k' \rVert}$

Dieses Verfahren ist die theoretische Grundlage für QR-Zerlegungen aber nummerisch überhaupt nicht stabil und wird deshalb auch nicht verwendet in der Praxis.

#### Householder-Spiegelungen

Eine Householder Spiegelung ist gegeben durch eine Matrix $Q = H(v) := I - 2\,\frac{v v^{H}}{v^{H} v}$

wobei 

$
v =
\begin{cases}
\frac{1}{2}\bigl(a + \lVert a \rVert_2\, e_1\bigr), & \text{falls } a_1 > 0, \\
\frac{1}{2}\bigl(a - \lVert a \rVert_2\, e_1\bigr), & \text{falls } a_1 \le 0.
\end{cases}
$

Diese Spiegelung bildet einen Vektor a auf $\lVert a \rVert _2\, e_1$ ab.

Somit können wir relativ leicht eine QR-Zerlegung erreichen, indem wir für alle Spalten $k$ $a = A_{k:m,\,k}$ setzen. Die entstandenen Householder-Matritzen ergeben dann $Q = Q_1 \cdots Q_{n-1}$ und die fertig bearbeitete Matrix A ist dann $R = Q_{n-1} \cdots Q_1\, A$

#### Givens-Drehungen

Für dünnbesetzte Matrizen kann es günstiger sein, die fehlenden Nullen in der Matrix schrittweise zu erstellen anstatt durch Householder-Spiegelungen. Dafür benützen wir Givens-Drehungen für die einzelnen "Q-Schritte":

```python
def givens(i,j,a):
    Q = np.eye(len(a))
    g = a[i]/np.sqrt(abs(a[i])**2 + abs(a[j]**2))
    s = a[j]/np.sqrt(abs(a[i])**2 + abs(a[j]**2))

    Q[i,i] = g
    Q[i, j] = s
    Q[j, i] = -s 
    Q[j, j] = g
    return Q
```

Für jede Spalte sind soviele Givens-Rotationen nötig, wie die Anzahl der Einträge, die nicht null sind.

### SVD

Sei $A \in \mathbb{C}^{m \times n}$ eine Matrix vom Rang $r$. Dann besitzt $A$ eine Singulärwertzerlegung $A = U,\Sigma,V^H$, wobei $U \in \mathbb{C}^{m \times m}$ und $V \in \mathbb{C}^{n \times n}$ unitär sind ($U^H U = I$, $V^H V = I$) und $\Sigma \in \mathbb{R}^{m \times n}$ eine (rechteckige) Diagonalmatrix ist mit $\Sigma = \operatorname{diag}(\sigma_1,\dots,\sigma_r,0,\dots)$ und $\sigma_1 \ge \sigma_2 \ge \dots \ge \sigma_r > 0$.

Die Singulärwerte erfüllen $\sigma_i^2 = \lambda_i(A^H A) = \lambda_i(A A^H)$. Der Rang ist $r = |{, i \mid \sigma_i > 0 ,}|$ (Anzahl positiver Singulärwerte).

Mit der Blockzerlegung $U = [,U_1\ U_2,]$, $V = [,V_1\ V_2,]$ und $\Sigma = \begin{pmatrix} \Sigma_r & 0 \ 0 & 0 \end{pmatrix}$ mit $\Sigma_r = \operatorname{diag}(\sigma_1,\dots,\sigma_r)$ gilt $A = U_1 \Sigma_r V_1^H$, wobei $U_1 \in \mathbb{C}^{m \times r}$ und $V_1 \in \mathbb{C}^{n \times r}$.

Die Moore-Penrose-Pseudoinverse ist $A^+ = V_1 \Sigma_r^{-1} U_1^H$. Für das Least-Squares-Problem $\min_x |Ax-b|_2$ ist die eindeutige Minimal-Norm-Lösung $x^* = A^+ b = V_1 \Sigma_r^{-1} U_1^H b$.

Normen und Kondition: $|A|_2 = \sigma_1$, $|A|_F = \sqrt{\sum_i \sigma_i^2}$, und für reguläres $A$ ist $cond(A) = \sigma_1/\sigma_r$.

Beste Rang-$k$-Approximation: $A_k = \sum_{i=1}^k \sigma_i u_i v_i^H$ (mit $u_i$, $v_i$ als $i$-te Spalten von $U$ bzw. $V$) minimiert den Fehler in $|\cdot|_2$ und $|\cdot|_F$ unter allen Matrizen vom Rang $\le k$.


## Ausgleichsrechnung

Grundidee der Auslgeichsrechnung: Bestmögliche Approximation von gegebenen Datenpunkten (z.B. Messdaten) durch ein Modell mittels Minimierung des Fehlers. ("Least Squares")

### Linear

Wenn die zugrundeliegende Funktion linear von ihren Parametern abhängt, sprechen wir von linearer Ausgleichsrechnung.

Wir versuchen den Fehler $\sum_{i=1}^{m} \lVert f(a, x_i) - y_i \rVert^2 = \lVert Ax-b \rVert _2$ zu minimieren.

Die Norm des **Residuums** $ \lVert b-Ax \rVert _2$ gibt uns den Fehler.

In pythonb können wir relativ simpel numpy.linalg.lstsq benutzen.

#### Normalengleichung

Um f(x) zu minimieren können wir die lineare Abbildung als Matrix A schreiben und die folgende Gleichung lösen: $A^HAx=A^Hb$

Diese Gleichung heist Normalengleichung. Der grosse Nachteil an dieser Methode ist, dass $A^HA$ sehr schlecht konditioniert ist. Deshlab empfiehlt es sich nicht diese Methode zu nutzen.

Wir können das ganze ein wenig Umgehen indem wir $A^HAx=A^Hb$ umformen zu $\begin{bmatrix}- I & A \\A^{H} & 0\end{bmatrix}\begin{bmatrix}r \\x\end{bmatrix}=\begin{bmatrix}b \\0\end{bmatrix}$

#### mit QR

Wir versuchen immer noch $\lVert r \rVert _2^2$ zu minimieren. Es ist bekannt dass orthogonale Matrizen die Norm unverändert lassen. Wir betrachten also die QR-Zerlegung.

$\lVert r \rVert _2^2$ = $\lVert R x - \tilde{b} \rVert _2^2$ wobei $\tilde{b} = Q^Hb$

Somit können wir die Lösung des systems durch $x = \tilde{R}^{-1}\,\tilde{b}_1$ bestimmen. Wobei $\tilde{R}^{-1}$ die reduzierte Form (ohne die 0-Zeilen am Ende) der Matrix R ist.

#### mit SVD

Die Normerhaltung unitärer/orthogonaler Matrizen liefert (wie vorher) folgende Umformung:
$\lVert A x - b \rVert _2 = \lVert U \Sigma V^{H} x - b \rVert _2$

Wir minimieren das Residuum mit $x = V_1 \Sigma_r^{+} U_1^{H} b$

dabei ist r der Rang von A, $V_1$ die ersten r Spalten von V, $\Sigma_r^{+}$ $\Sigma$ r positiven Singulärwerte und $U_1^{H}$ die r ersten Zeilen von $U^H$.

Professionelle Lösungen wie numpy.linalg.lstsq benutzen auch diese Methode.

($V_1 \Sigma_r^{+} U_1^{H}$ bezeichnen wir als Pseudoinverse der Matrix A.)

#### Totale Ausgleichsrechnung

Was wenn die Messfehler und A und b verhindern dass b im Bild von A liegt (und es somit keine Lösung gibt)?

Dann minimieren wir $\lVert [A\ b] - [\hat{A}\ \hat{b}] \rVert _F$ wobei $\hat{A}\hat{x} = \hat{b}$ möglichst nahe am ursrpünglichen System liegt.

Wir definiren C = [A b] und $\hat{C} = [\hat{A}\ \hat{b}]$

Das Optimum findet sich bei $\hat{C} = \sum_{j=1}^{n} \sigma_j\, u_j v_j^{H}$

Falls $v_{n+1,n+1} \neq 0$ dann haben wir die Lösung $\hat{x} = v_{n+1}/v_{n+1,n+1}$

#### mit linearen Nebenbedingungen

##### Lagrange-Multiplikatoren (Standardmethode)

Wir setzten nun voraus dass $Cx = d$ (z.B. durch Physikalische Gesetze)

Dann können wir folgendes Gelichungssystem lösen:
$\begin{bmatrix} A^{T}A & C^{T} \\ C & 0 \end{bmatrix} \begin{bmatrix} x \\ m \end{bmatrix} = \begin{bmatrix} A^{T} b \\ d \end{bmatrix}$

Nummerisch schöner lösen, können wir das ganze mit einer Block-LU Zerlegung:

$\begin{bmatrix} A^{T}A & C^{T} \\ C & 0 \end{bmatrix} = \begin{bmatrix} R^{T} & 0 \\ G & S \end{bmatrix} \begin{bmatrix} R & G^{T} \\ 0 & S^{T} \end{bmatrix}$

wobei
- $A^TA = R^TR$ (Cholesky-Zerlegung)
- $C = GR$ (LU-Zerlegung)
- $-GG^T = S^TS$ (Cholesky-Zerlegung)


##### Eliminationsmethode

"Elimination parametrisiert die Nebenbedingungen vollständig und reduziert das Problem auf ein gewöhnliches Least-Squares im Kern von C."

Berechnung:
1. SVD von C:
$C = U \begin{bmatrix} \Sigma & 0 \end{bmatrix} \begin{bmatrix} V_1^{T} \\ V_2^{T} \end{bmatrix}$
2. Partikuläre Lösung von $Cx = d$:
$x_0 = V_1 \Sigma^{-1} U^{T} d$
3. Anwendung von Least Squares auf:
$\lVert A V_2 y - (b - A x_0) \rVert _2^2$

### Nicht-Linear

"Wenn das zugrunde liegende Modell f¨ur das Ausgleichsproblem nicht linear in den Parametern ist, dann reden wir von nichtlinearer Ausgleichsrechnung."

Wir wollen x finden sodass $x = \arg\min_{u \in \mathbb{R}^n} \left( \sum_{i=1}^{m} \lvert f(t_i, u) - y_i \rvert^2 \right) = \arg\min_{u \in \mathbb{R}^n} \frac{1}{2} \lVert F(u) \rVert _2^2$

Wobei die Residuenfunktion wie folgt definiert ist:
$F(x) = \begin{bmatrix} f(t_1,x)-y_1 \\ \vdots \\ f(t_m,x)-y_m \end{bmatrix}$


#### Newton

Wir suchen eine Methode, mit der wir $\phi(x) = \frac{1}{2} \lVert F(u) \rVert _2^2$ minimieren können.

Das geht mit dem iterativen Newton-Verfahren:

$x_{k+1} = x_k - (H_{\phi}(x)^{-1})* grad \phi(x)$

Wir berechnen pro iteration einmal die Hesse matrix aber nicht direkt die Inverse sondern lösen jeweils das Gleichungssystem $H_{\Phi}(x_k)\, s = grad \Phi(x_k)$ und setzen $x{k+1} = x_k -s$

#### Gauss-Newton

Das Gauss-Newton Verfahren beruht auf dem Newton-Verfahren und vereinfacht es durch Vernachlässigung der zweiten Ableitung. Ausserdem ist reines Newton-Verfahren nicht immer möglich, da F(x) nicht immer zweimal stetig differenzierbar ist. 

Der Iterationsschritt sieht dann folgendermassen aus:

$x^{(k+1)} = x^{(k)} - s \quad \text{mit} \quad s := \arg\min_{z \in \mathbb{R}^n} \lVert F(x^{(k)}) - DF(x^{(k)})\, z \rVert _2^2$

#### BFGS, GD, SGD, RCD, LM, ADAM

- **BFGS**: 
Anstelle einer Linearisierung des Modells in jedem Schritt wie bei Gauss-Newton verwenden wir hier eine schnelle Variante von der Broyden Methode.

- **GD**:
Update in richtung des steilsten Abstiegs (Gradient Descent). $s = \lambda_kD\phi(x_k)$

- **SGD**:
Stochastische Variante des GD

- **RCD**:
Zufälliger Koordinatenabstieg

- **ADAM**
adaptives, momentbasiertes SGD-Verfahren
